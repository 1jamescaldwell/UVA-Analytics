Trying to determine students who have lost VA residency for tuition purposes


```python
import numpy as np 
import pandas as pd 

df = pd.read_excel(r"...771 Enrollment - Tuition Residency.xlsx")

state_change = df.pivot(
    index='Student System ID',
    columns=['Term Desc'],
    values='Home Address State'
).reset_index()


state_change_res = df.pivot(
    index='Student System ID',
    columns=['Term Desc'],
    # values='Student Residency'
    values='Tuition Residency'
).reset_index()

state_change_res=state_change_res.rename(columns={"2026 Fall": "2026 Fall Residency","2026 Spring": "2026 Spring Residency"})

state_change['Term Desc'] = '2026 Fall'
state_change_res['Term Desc'] = '2026 Fall' 

# Add back name and other helpful columns
state_change = state_change.merge(df, 
                                  how='left',
                                  on=['Student System ID','Term Desc'])

# Add residency by term columns
state_change = state_change.merge(state_change_res, 
                                  how='left',
                                  on=['Student System ID'])

# Filters
state_change_filtered = state_change[
    # (state_change['2026 Fall Residency'] == 'In-State') & # Use this line for Student Residency Field
    (state_change['2026 Fall Residency'] == 'V') & # Use this line if using Tuitino Residency Field
    (state_change['2026 Fall'] != state_change['2026 Spring']) &
    (state_change['2026 Fall'].notna()) &
    (state_change['2026 Spring'].notna()) &
    (state_change['2026 Fall'] != 'VA')
]

state_change_filtered.to_excel('output.xlsx',sheet_name='enrollment')
```
